# Ξ–χ Scalar-Entropy Framework (Model 5: Horndeski Extension)

This notebook lets you simulate and visualize a scalar-tensor entropy-curvature framework from first principles. Based on:

- Recursive entropy field $\Xi(x, t)$
- Curvature feedback via Horndeski $G_5(\Xi)$ coupling
- Birefringence, entropy loops, and quantum testability

## Simulate Model 5: Scalar-Entropy + Curvature Feedback

```python
def laplacian(field):
    return np.gradient(np.gradient(field, axis=0), axis=0) + np.gradient(np.gradient(field, axis=1), axis=1)

def rk4_step_model5(xi, chi, dt, lambda_coupling, gamma_5, m):
    def rhs(xi, chi):
        lap_xi = laplacian(xi)
        lap_chi = laplacian(chi)
        G5 = gamma_5 * xi**m
        return lap_xi + lambda_coupling * xi * chi + G5 * lap_chi

    k1 = rhs(xi, chi)
    k2 = rhs(xi + 0.5 * dt * k1, chi)
    k3 = rhs(xi + 0.5 * dt * k2, chi)
    k4 = rhs(xi + dt * k3, chi)
    return xi + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
```

### Run Simulation

```python
steps = 100
size = 100
xi = np.random.normal(0.5, 0.1, (size, size))
chi = np.random.normal(0.0, 0.01, (size, size))

xi_snapshots = []
entropy_series = []
for _ in range(steps):
    xi = rk4_step_model5(xi, chi, dt=0.01, lambda_coupling=0.3, gamma_5=0.05, m=1.0)
    xi_snapshots.append(xi.copy())
    p = np.abs(xi.flatten()) / np.sum(np.abs(xi.flatten()))
    p = p[p > 0]
    entropy_series.append(-np.sum(p * np.log(p)))
```

### Visualize Last Ξ Field

```python
plt.figure(figsize=(6, 6))
plt.imshow(xi_snapshots[-1], cmap='viridis', origin='lower')
plt.colorbar(label='Ξ')
plt.title('Ξ Field Snapshot')
plt.show()
```

### Plot Entropy $S(t)$

```python
plt.plot(entropy_series)
plt.title('Entropy S(t)')
plt.xlabel('Step')
plt.ylabel('Entropy')
plt.grid(True)
plt.show()
```
